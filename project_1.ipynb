# The finger length of each people

1. Pandas method

```python
import pandas as pd


df = pd.read_csv("MacdonellDF.csv")


print(f"Total rows: {len(df)}")
display(df.head())


finger_mean = df["finger"].mean()
finger_median = df["finger"].median()
finger_mode = df["finger"].mode()[0]  

print(f"Mean:   {finger_mean:.4f}")
print(f"Median: {finger_median:.4f}")
print(f"Mode:   {finger_mode:.4f}")
```

2. The hard way

```python
import csv

values = []
counts = {}  

with open("MacdonellDF.csv", mode="r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        val = float(row["finger"])
        values.append(val)
        counts[val] = counts.get(val, 0) + 1


n = len(values)
hard_mean = sum(values) / n


sorted_values = sorted(values)
mid = n // 2
if n % 2 == 1:
    hard_median = sorted_values[mid]
else:
    hard_median = (sorted_values[mid - 1] + sorted_values[mid]) / 2

hard_mode = None
max_count = 0
for val, count in counts.items():
    if count > max_count:
        max_count = count
        hard_mode = val

print(f"Hard way Mean:   {hard_mean:.4f}")
print(f"Hard way Median: {hard_median:.4f}")
print(f"Hard way Mode:   {hard_mode:.4f}")


```

Data visualization

```python

finger_counts = df["finger"].value_counts().sort_index()

print("Distribution of Left Middle Finger Length (each * = 5 people)\n")

for f_val, count in finger_counts.items():
    stars = "*" * (count // 5)
    print(f"{f_val:4.1f} cm | {stars} ({count})")
```
