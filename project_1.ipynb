# #Finger length in Macdonell's 1902 sample

Macdonell (1902) measured 3,000 men serving sentences in England and Wales,
recording their height and the length of their left middle finger. I use the
finger column to describe the sample: where it centres, and where the values
are concentrated.

- Data: `HistData::MacdonellDF`, from W. R. Macdonell (1902), "On Criminal
  Anthropology and the Identification of Criminals", *Biometrika* 1(2), 177–227.
  [Documentation](https://vincentarelbundock.github.io/Rdatasets/doc/HistData/MacdonellDF.html)
- One row = one man
- Numeric column: `finger`, left middle finger length in cm (values run 9.5 to 13.5)
- 3,000 rows, no missing values


```python
import pandas as pd

CSV_PATH = "MacdonellDF.csv"
COLUMN = "finger"

df = pd.read_csv(CSV_PATH)

print(f"Rows: {len(df):,}")
print(f"Columns: {list(df.columns)}")
print()
print(df.head())

values = df[COLUMN].dropna()

finger_mean = values.mean()
finger_median = values.median()
finger_mode = values.mode().iloc[0]      

print()
print(f"n      = {len(values):,}")
print(f"mean   = {finger_mean:.4f} cm")
print(f"median = {finger_median:.4f} cm")
print(f"mode   = {finger_mode:.4f} cm ({int((values == finger_mode).sum())} men)")
```

The mean (11.55 cm) sits slightly above the median (11.50 cm), which means the
distribution has a thin right tail.

The mode is 11.5 cm and it is reached by 233 men — about 8% of the sample.


## The same thing without pandas

Everything below re-reads the file with the standard library only, and the
three statistics are computed by hand. 

```python
import csv

values = []
counts = {}  

with open("MacdonellDF.csv", mode="r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        val = float(row["finger"])
        values.append(val)
        counts[val] = counts.get(val, 0) + 1


n = len(values)
hard_mean = sum(values) / n


sorted_values = sorted(values)
mid = n // 2
if n % 2 == 1:
    hard_median = sorted_values[mid]
else:
    hard_median = (sorted_values[mid - 1] + sorted_values[mid]) / 2

hard_mode = None
max_count = 0
for val, count in counts.items():
    if count > max_count:
        max_count = count
        hard_mode = val

print(f"Hard way Mean:   {hard_mean:.4f}")
print(f"Hard way Median: {hard_median:.4f}")
print(f"Hard way Mode:   {hard_mode:.4f}")


```

## Drawing the distribution

The counts below come from the dictionary built above, so the drawing code
uses nothing but the standard library

```python

bins = [9.5, 10.0, 10.5, 11.0, 11.5, 12.0, 12.5, 13.0, 13.5]
binned_counts = pd.cut(df["finger"], bins=bins, include_lowest=True).value_counts().sort_index()

print("Left Middle Finger Length Histogram (each █ = 25 people)\n")

for interval, count in binned_counts.items():
    bars = "█" * (count // 25)
    print(f"{str(interval):13s} | {bars} ({count})")
```

## What this shows

The distribution is centred at 11.5 cm and it is symmetric. The right tail is close to the left. We assume the data was randomly selected.
